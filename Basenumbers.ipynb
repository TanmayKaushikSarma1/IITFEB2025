# Functions

```python
x = 2756
result = 0
multiplier = 1
mybase = 8
```

```python
def extract_last_digit():
    global x
    global result
    global multiplier
    print("Number:" , x)
    d = x % 10
    print("Last Digit:" , d)
    print("Multiplier:" , multiplier)
    result += d*multiplier
    print("Result:" , result)
    x = x // 10
    multiplier *= mybase
    print("New Number:" , x)
    print("New Multiplier:" , multiplier)
```
