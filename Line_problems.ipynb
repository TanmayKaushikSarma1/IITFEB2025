# Create a condition on x1,x2 and x where print True if x is on the line joining x1 and x2

```python
x1 = 5
x2 = 9
x = 8
```

```python
def line_problems(x1 , x2 , x):
    if (x >= x1 and x <= x2) or (x >= x2 and x <= x1):
        return True
    else:
        return False
```

```python
result = line_problems(x1 , x2 , x) 
print("Result:" , result)
```
